```python
# ============================================================
# TITANIC DATASET - SIMPLE DATA VISUALIZATION
# ============================================================

# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# 2. Load the already uploaded dataset
# FILE ADDRESS AND FILE NAME ARE NOT CHANGED
CSV_PATH = "/content/titanic (1).csv"

df = pd.read_csv(CSV_PATH)


# 3. Remove accidental spaces from column names
df.columns = df.columns.str.strip()


# 4. Select the required input columns
INPUT_COLUMNS = [
    "pclass",
    "sex",
    "age",
    "sibsp",
    "parch",
    "fare",
    "embarked"
]

TARGET_COLUMN = "survived"


# 5. Create a DataFrame containing inputs and target
data = df[INPUT_COLUMNS + [TARGET_COLUMN]].copy()


# 6. Give readable names to survival classes
data["survival_status"] = data["survived"].map({
    0: "Not Survived",
    1: "Survived"
})


# 7. Create family-size column
# Family size = passenger + siblings/spouse + parents/children
data["family_size"] = data["sibsp"] + data["parch"] + 1


# 8. Display basic information
print("Columns Used for Visualization:")
print(INPUT_COLUMNS)

print("\nTarget Column:")
print(TARGET_COLUMN)

print("\nFirst Five Rows:")
print(data.head())


# ============================================================
# VISUALIZATION 1: MISSING VALUES
# ============================================================

missing_values = data[INPUT_COLUMNS].isnull().sum()

plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing_values.index,
    y=missing_values.values
)

plt.title("Missing Values in Titanic Input Columns")
plt.xlabel("Input Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 2: SURVIVAL CLASS DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=data,
    x="survival_status"
)

plt.title("Titanic Survival Class Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 3: SURVIVAL BY PASSENGER CLASS
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=data,
    x="pclass",
    hue="survival_status"
)

plt.title("Survival Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survival Status")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 4: SURVIVAL BY GENDER
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=data,
    x="sex",
    hue="survival_status"
)

plt.title("Survival Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survival Status")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 5: AGE DISTRIBUTION BY SURVIVAL
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=data,
    x="age",
    hue="survival_status",
    bins=30,
    kde=True,
    multiple="stack"
)

plt.title("Age Distribution Based on Survival")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 6: FARE BY SURVIVAL STATUS
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="survival_status",
    y="fare"
)

plt.title("Fare Distribution by Survival Status")
plt.xlabel("Survival Status")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 7: SURVIVAL BY EMBARKATION PORT
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=data,
    x="embarked",
    hue="survival_status"
)

plt.title("Survival Distribution by Embarkation Port")
plt.xlabel("Embarkation Port")
plt.ylabel("Number of Passengers")
plt.legend(title="Survival Status")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 8: SURVIVAL BY FAMILY SIZE
# ============================================================

plt.figure(figsize=(10, 5))

sns.countplot(
    data=data,
    x="family_size",
    hue="survival_status"
)

plt.title("Survival Distribution by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Number of Passengers")
plt.legend(title="Survival Status")
plt.tight_layout()
plt.show()


# ============================================================
# VISUALIZATION 9: CORRELATION HEATMAP
# ============================================================

numeric_columns = [
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare",
    "family_size",
    "survived"
]

correlation_matrix = data[numeric_columns].corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()


print("\nAll Titanic visualizations were generated successfully.")
```
